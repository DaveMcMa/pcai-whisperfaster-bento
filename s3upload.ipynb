```python
import json
from datetime import datetime

def pretty_s3_response(contents):
    if not contents:
        print("No files found in bucket.")
        return
    print(f"\n{'File':<40} {'Size':>10}  {'Last Modified'}")
    print("-" * 70)
    for obj in contents:
        size_kb = obj['Size'] / 1024
        modified = obj['LastModified'].strftime('%Y-%m-%d %H:%M')
        print(f"{obj['Key']:<40} {size_kb:>8.1f}KB  {modified}")

# Upload the bento file
s3.upload_file('whisper-faster.bento', 'rundmc', 'whisper-faster-v6.bento')

# Verify upload
response = s3.list_objects_v2(Bucket='rundmc')
pretty_s3_response(response.get('Contents', []))
```
